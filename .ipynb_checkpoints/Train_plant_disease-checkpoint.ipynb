```python
### Training Image Preprocessing
```

```python
training_set = tf.keras.utils.image_dataset_from_directory(
    "train",
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    verbose=True,
)
```

```python
###Validation Image Pre Processing
```

```python
validation_set = tf.keras.utils.image_dataset_from_directory(
    "valid",
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    verbose=True,
)
```

```python
training_set
```

```python
for x,y in training_set:
    print(x,x.shape)
    print(y,y.shape)
```

```python
##Building Models
```

```python
from tensorflow.keras.layers import Dense, Conv2D, MaxPool2D, Flatten, Dropout
from tensorflow.keras.models import Sequential
```

```python
models = Sequential()
```

```python
##Building Convolution Layer
```

```python
models.add(Conv2D(filters=32,kernel_size=3, padding='same',activation='relu', input_shape=[128,128,3]))
models.add(Conv2D(filters=32,kernel_size=3,activation='relu'))
models.add(MaxPool2D(pool_size=2,strides=2))
```

```python
models.add(Conv2D(filters=64,kernel_size=3, padding='same',activation='relu'))
models.add(Conv2D(filters=64,kernel_size=3, padding='same',activation='relu'))
models.add(MaxPool2D(pool_size=2,strides=2))
```

```python
models.add(Conv2D(filters=128,kernel_size=3, padding='same',activation='relu'))
models.add(Conv2D(filters=128,kernel_size=3, padding='same',activation='relu'))
models.add(MaxPool2D(pool_size=2,strides=2))
```

```python
models.add(Conv2D(filters=256,kernel_size=3, padding='same',activation='relu'))
models.add(Conv2D(filters=256,kernel_size=3, padding='same',activation='relu'))
models.add(MaxPool2D(pool_size=2,strides=2))
```

```python
models.add(Conv2D(filters=512,kernel_size=3, padding='same',activation='relu'))
models.add(Conv2D(filters=512,kernel_size=3, padding='same',activation='relu'))
models.add(MaxPool2D(pool_size=2,strides=2))
```

```python
models.add(Dropout(0.25))
```

```python
models.add(Flatten())
```

```python
models.add(Dense(units=1500,activation='relu'))
```

```python
models.add(Dropout(0.40))
```

```python
#Output layer
models.add(Dense(units=38,activation='softmax'))
```

```python
##Compiling layers
```

```python
models.compile(optimizer=tf.keras.optimizers.Adam(
    learning_rate=0.0001),loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
models.summary()
```

### Training Model

```python
training_history = models.fit(x=training_set,validation_data=validation_set,epochs=10)
```
